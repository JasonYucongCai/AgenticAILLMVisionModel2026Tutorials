## Single Object Tracking

This document gives a basic introduction to **Single Object Tracking (SOT)**.

Unlike Object Detection, the input to SOT consists of:

- one **reference frame**, and
- a time series of 3D images that need to be tracked.

Common datasets include **OTB-100**, **GOT-10k**, and **LaSOT**.

- **OTB-100** is one of the earliest classic benchmarks. It is difficult to access, its original website is no longer available, and the Internet Archive cannot be reliably accessed from ..... or locally.
- **GOT-10k** is still available, but registration is required.
- **LaSOT** can still be accessed from the Hugging Face website, but it is not available on the mirror service in ......

## Two Approaches

## Method 1: Modular Neural Network

A modular neural network is similar to Agentic AI, but it is more commonly associated with first-generation AI techniques as a way of building tools. Agentic AI, by contrast, is more often associated with second-generation AI techniques and typically runs its workflow in a loop.

Its core idea is to combine multiple neural networks with known functions to achieve a specific goal.

The main advantages of a modular neural network are:

- low cost,
- controllability, and
- ease of fast industrial deployment.

Its components can be deployed in a distributed manner and can support partial version control. In practice, however, it is often affected by convergence speed, computational speed, functional consistency, and accuracy.

## Modules Used

The module design in this document uses:

- a **Faster R-CNN**,
- a **ResNet**, and
- a **Vision Transformer**.

## Method 1 Workflow

First, select a subset of random data from the training dataset and fine-tune a Faster R-CNN, referred to as **model A**.

Model A can serve as the **Regional Proposal Network** for the image to be tracked. After fine-tuning, model A remains frozen and is not updated further during later training.

Next, use a **ResNet** as the **feature extraction network** for the reference frame, referred to as **model B**. Its output is passed directly to the Vision Transformer (ViT).

At the same time, a **skip connection** sends the image to be tracked directly to a **projector** that crops it into patches, just like the original ViT process.

The ViT then receives:

- the reference-frame features from **model B**,
- the patches from the original reference frame through the skip connection, and
- the regional proposal position parameters from **model A** for the tracking image, placed after the CLS token.

These components form a **Unified Network**. The modified ViT and the parameters of model B need to be trained and fine-tuned, thereby forming a neural network for Single Object Tracking.

## Method 2: ViT with RoPE

Using ViT with RoPE is a very natural approach and has also been widely used in recent SOT neural networks.

Its core idea is to use **Rotary Positional Embedding (RoPE)**. One benefit is that the transformer input sequence can support variable length, unlike absolute positional embedding.

RoPE also changes token relationships into **relative positions**, which significantly strengthens correlation and contextual association. The problem addressed here is Single Object Tracking, which involves the projection of an input frame and a reference frame, that is, contextual association between image tokens. Therefore, RoPE is more suitable than absolute positional embedding, and it can also handle variable input lengths caused by different image sizes.

## Method 2 Workflow

The method is:

1. Download a pretrained **ViT with RoPE**.
2. Modify the projector so that the reference frame and the image frame are cropped to the same size through the same projector.
3. Add a **CLS token**.
4. Freeze all parameters.
5. Set the initial projector, the first transformer layer, one middle transformer layer, and the final output layer to trainable mode.
6. Train for one epoch first. This is a fast trial-and-error step, different from layer-wise unfreezing.
7. Then set the entire model to trainable mode and continue training.

This completes the goal of Single Object Tracking.
