## 2. Load your API key from a `.env` file and call the model API

After obtaining your API key as shown earlier and saving it in a `.env` file, you can call the model API.

### DeepSeek

````python
import os
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent, Runner, OpenAIChatCompletionsModel
from agents import set_tracing_disabled

load_dotenv(override=True)

# Disable tracing (not needed for DeepSeek)
set_tracing_disabled(True)
# OPENAI_AGENTS_DISABLE_TRACING=1

# 1. Configure DeepSeek client
client = AsyncOpenAI(
    api_key=os.getenv('DEEPSEEK_API_KEY'),   # Your DeepSeek API key
    base_url="https://api.deepseek.com/v1"
)
````

### .....

````python
# Initialize ..... client
client = AsyncOpenAI(
    api_key=os.getenv("....._API_KEY"),
    base_url="https://api........./api/llm/v1"
)
````

Note: unlike the previous article, this version uses `AsyncOpenAI`. This feature is non-blocking, so it can simulate a multithreaded style without creating extra threads, which makes it easier to handle API calls with high I/O latency.

## 3. Use `OpenAIChatCompletionsModel` to stay consistent with GPT-style calls

To stay consistent with OpenAI GPT-series model calls, use `OpenAIChatCompletionsModel` to bridge into the OpenAI SDK ecosystem.

In this ecosystem, model inputs and outputs use a specific JSON format, and native features such as `function_call` are supported. This architecture has already become an international standard. Mainstream large models in China, including DeepSeek, Minimax, Moonshot AI (Kimi), Zhipu AI (GLM), ByteDance (Doubao), and Baidu (ERNIE), all fully support these capabilities and use them as a unified standard interface.

This has become the foundational architecture for current Agentic AI systems and large-model services in China, and there is currently no replacement requirement. It is relatively stable and can be used with confidence.

### DeepSeek

````python
# 2. Wrap it with the bridge class
model = OpenAIChatCompletionsModel(
    model="deepseek-chat",   # or "deepseek-reasoner" (note: reasoner doesn't support tools)
    openai_client=client
)
````

### .....

````python
# 2. Wrap it with the bridge class
model = OpenAIChatCompletionsModel(
    model="DeepSeek-R1-0528",   # or "Qwen3-30B-A3B" etc.
    openai_client=client
)
````

## 4. Define the Agent, whose behavior is controlled by `instruction` (that is, `system`)

### DeepSeek example

````python
# 3. Create your agent with this model
agent = Agent(
    name="Albert Einstein",
    instructions="You are an insightful academician.",   # fixed typo
    model=model,
)
````

## 5. Define an async function and use `await` to let the AI Agent handle a task

### DeepSeek example

````python
# 4. Define the async function
async def main():
    result = await Runner.run(agent, "Explain the theory of relativity in simple terms.")
    print(result.final_output)

# 5. In Jupyter, directly await the main function
await main()
````
